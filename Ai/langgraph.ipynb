```python
from langchain.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.runnables import RunnableSequence
import os
from dotenv import load_dotenv
```

```python
load_dotenv()

api_key = os.getenv("GOOGLE_API_KEY")
```

```python

llm = ChatGoogleGenerativeAI(
    model="models/gemini-1.5-flash",
    google_api_key=os.getenv("GOOGLE_API_KEY"),
    temperature=2.0,
    max_output_tokens=1024,
    top_p=0.8,
)
```

```python

# Define prompt template
template = PromptTemplate.from_template("Write a story with character, plot, problem, climax and solutions using the following words: {text} Note: Keep the story descriptive and engaging. The story should be at least 1000 words long and should have a clear beginning, middle, and end. The story should also have a clear theme or message that is conveyed through the characters and plot. The story should be written in a way that is easy to understand and follow, with clear transitions between scenes and events. The story should also have a clear resolution to the problem presented in the plot, with the characters learning and growing from their experiences.")
```

```python
chain = RunnableSequence(template, llm, StrOutputParser())
```

```python
output = (chain.invoke({"text": "fish dajjal fuji coldplay f*ck"}))
```

```python
with open("output.txt", "w") as f:
    f.write(output)
```

```python
from langchain.memory import ConversationBufferMemory
from langchain.chains import ConversationChain

memory = ConversationBufferMemory()
chat = ConversationChain(llm=llm, memory=memory)

print(chat.invoke("My name is Abdullah"))
print(chat.invoke("What is my name?"))
```
